```python
import sys
sys.path.insert(0, '.')
import os
import whisperx
os.environ["HF_TOKEN"] = os.getenv("HF_TOKEN", "")

from modules.stt import transcribe, transcribe_with_speakers, format_dialogue
from modules.logger import setup_logger

device = "cuda" if __import__('torch').cuda.is_available() else "cpu"
audio_path = "data/input/test.mp3"
print(f"✅ 导入完成 | device={device} | audio={audio_path}")
```

## Step 1: 纯转录

```python
%%time
model = whisperx.load_model("base", device)
audio = whisperx.load_audio(audio_path)
result_transcribe = model.transcribe(audio, language="zh")

plain_text = transcribe(audio_path, "base")
print(f"片段数: {len(result_transcribe['segments'])}")
print(f"总字符: {len(plain_text)}")
print()
print("--- 转录文本（纯文本）---")
print(plain_text)
```

## Step 2: 时间轴对齐

```python
%%time
align_model, align_metadata = whisperx.load_align_model(language_code="zh", device=device)
result_aligned = whisperx.align(result_transcribe["segments"], align_model, align_metadata, audio, device)

for i, seg in enumerate(result_aligned["segments"]):
    t = f"[{seg['start']:.1f}s -> {seg['end']:.1f}s]"
    print(f"{t:30s} {seg['text'].strip()[:80]}")
```

## Step 3: 说话人分离 + 合并

需要 HF_TOKEN 已授权 pyannote 相关模型。

```python
%%time
import torch
import torchaudio
import pandas as pd

from modules.stt import _get_diarization_pipeline

# 用 torchaudio 预加载音频到内存，绕过 pyannote 的 AudioDecoder 依赖
waveform, sample_rate = torchaudio.load(audio_path)
audio_in_memory = {"waveform": waveform, "sample_rate": sample_rate}
print(f"音频: {waveform.shape[1]/sample_rate:.0f}s, {sample_rate}Hz, {waveform.shape[0]}ch")

pipeline = _get_diarization_pipeline()
diarize_output = pipeline(audio_in_memory, num_speakers=2)

# pyannote 4.x → DiarizeOutput → Annotation → DataFrame（whisperx 需要）
anno = diarize_output.exclusive_speaker_diarization
df_rows = []
for turn, _, speaker in anno.itertracks(yield_label=True):
    df_rows.append({"start": turn.start, "end": turn.end, "speaker": speaker})
diarization_df = pd.DataFrame(df_rows)
print(f"说话人片段数: {len(diarization_df)}")

result_diarized = whisperx.assign_word_speakers(diarization_df, result_aligned)

speakers = set()
for seg in result_diarized["segments"]:
    spk = seg.get("speaker", "UNKNOWN")
    speakers.add(spk)
    t = f"[{spk}] [{seg['start']:.1f}s->{seg['end']:.1f}s]"
    print(f"{t:40s} {seg['text'].strip()[:80]}")
print(f"\n识别到 {len(speakers)} 位说话人: {sorted(speakers)}")
```

## Step 5: 保存中间结果

供模块 2（对话拆解）、模块 3（问题诊断）直接加载，无需重跑 STT。

```python
import json
from pathlib import Path

out_dir = Path("data/output/intermediate")
out_dir.mkdir(parents=True, exist_ok=True)

# 保存 segments（结构化数据，模块2/3的核心输入）
with open(out_dir / "segments.json", "w", encoding="utf-8") as f:
    json.dump(segments, f, ensure_ascii=False, indent=2)

# 保存纯对话文本（方便快速阅读）
with open(out_dir / "dialogue.txt", "w", encoding="utf-8") as f:
    f.write(dialogue)

# 保存各说话人的话（按角色聚合）
teacher_lines = [s for s in segments if s["speaker"] == "SPEAKER_01"]
parent_lines = [s for s in segments if s["speaker"] == "SPEAKER_00"]
with open(out_dir / "by_role.json", "w", encoding="utf-8") as f:
    json.dump({
        "teacher": [{"text": s["text"], "start": s["start"], "end": s["end"]} for s in teacher_lines],
        "parent": [{"text": s["text"], "start": s["start"], "end": s["end"]} for s in parent_lines],
    }, f, ensure_ascii=False, indent=2)

print(f"已保存到 {out_dir}/")
print(f"  segments.json   — {len(segments)} 段，带说话人标签")
print(f"  dialogue.txt    — 格式化对话文本")
print(f"  by_role.json    — 老师 {len(teacher_lines)} 段 + 家长 {len(parent_lines)} 段")
```
